````python
# NEW APPROACH FOR WIDE FORMAT DATA
print("=== PROCESSING WIDE FORMAT TIME SERIES DATA ===")

def process_wide_format_data(df):
    """Process the wide format data where columns are time periods"""
    
    print("1. IDENTIFYING TIME PERIOD COLUMNS")
    print("=" * 50)
    
    # Extract all column names that contain dates
    time_period_columns = []
    for col in df.columns:
        col_str = str(col)
        # Look for columns with year patterns (2018-2025)
        if any(str(year) in col_str for year in range(2018, 2026)):
            time_period_columns.append(col)
    
    print(f"Found {len(time_period_columns)} time period columns")
    print("Sample time period columns:")
    for col in time_period_columns[:10]:
        print(f"  - {col}")
    
    return time_period_columns

time_period_cols = process_wide_format_data(df)
````

````python
# EXTRACT FUND INFORMATION AND CREATE PROPER STRUCTURE
print("\n2. EXTRACTING FUND METADATA")
print("=" * 50)

def extract_fund_metadata(df):
    """Extract fund manager and fund name information"""
    
    # Find the data starting point - look for rows with actual fund data
    data_start_idx = None
    fund_metadata = []
    
    for idx in range(len(df)):
        row = df.iloc[idx]
        row_str = str(row.iloc[1]) if len(row) > 1 else ""
        
        # Look for rows that contain fund manager names (not headers)
        if (pd.notna(row.iloc[1]) and 
            'fund' not in str(row.iloc[1]).lower() and 
            'manager' not in str(row.iloc[1]).lower() and
            'name' not in str(row.iloc[1]).lower() and
            str(row.iloc[1]).strip() != '' and
            str(row.iloc[1]).strip() != 'NaN'):
            
            if data_start_idx is None:
                data_start_idx = idx
            
            # Extract fund information
            fund_info = {
                'row_index': idx,
                'fund_manager': row.iloc[1] if len(row) > 1 else None,
                'fund_name': row.iloc[2] if len(row) > 2 else None,
                's_no': row.iloc[0] if len(row) > 0 else None
            }
            fund_metadata.append(fund_info)
    
    print(f"Data starts at row {data_start_idx}")
    print(f"Found {len(fund_metadata)} potential fund records")
    
    # Create fund metadata dataframe
    fund_meta_df = pd.DataFrame(fund_metadata)
    
    # Display sample
    if len(fund_meta_df) > 0:
        print("\nSample fund metadata:")
        print(fund_meta_df.head(10))
    
    return fund_meta_df, data_start_idx

fund_meta_df, data_start_idx = extract_fund_metadata(df)
````

````python
# CREATE TIME SERIES DATASET
print("\n3. CREATING TIME SERIES DATASET")
print("=" * 50)

def create_timeseries_dataset(df, fund_meta_df, time_period_cols, data_start_idx):
    """Convert wide format to long format time series"""
    
    time_series_data = []
    
    print("Processing time series data...")
    
    for _, fund_row in fund_meta_df.iterrows():
        row_idx = fund_row['row_index']
        fund_manager = fund_row['fund_manager']
        fund_name = fund_row['fund_name']
        
        # Extract NAV values across all time periods for this fund
        for time_col in time_period_cols:
            nav_value = df.at[row_idx, time_col]
            
            if pd.notna(nav_value) and isinstance(nav_value, (int, float)):
                # Extract date from column name
                date_str = extract_date_from_column(time_col)
                
                time_series_data.append({
                    'fund_manager': fund_manager,
                    'fund_name': fund_name,
                    'date_period': time_col,
                    'date': date_str,
                    'nav_value': nav_value,
                    'source_row': row_idx
                })
    
    time_series_df = pd.DataFrame(time_series_data)
    
    print(f"Created time series with {len(time_series_df)} NAV observations")
    print(f"Covering {time_series_df['fund_name'].nunique()} unique funds")
    
    if len(time_series_df) > 0:
        print("\nTime series sample:")
        print(time_series_df.head())
    
    return time_series_df

def extract_date_from_column(col_name):
    """Extract date from column name string"""
    import re
    from datetime import datetime
    
    col_str = str(col_name).upper()
    
    # Month mapping
    month_map = {
        'JANUARY': 1, 'JAN': 1, 'FEBRUARY': 2, 'FEB': 2, 'MARCH': 3, 'MAR': 3,
        'APRIL': 4, 'APR': 4, 'MAY': 5, 'JUNE': 6, 'JUN': 6, 'JULY': 7, 'JUL': 7,
        'AUGUST': 8, 'AUG': 8, 'SEPTEMBER': 9, 'SEP': 9, 'OCTOBER': 10, 'OCT': 10,
        'NOVEMBER': 11, 'NOV': 11, 'DECEMBER': 12, 'DEC': 12
    }
    
    # Try to extract year
    year_match = re.search(r'20(1[8-9]|2[0-5])', col_str)
    year = int(year_match.group()) if year_match else None
    
    # Try to extract month
    month = None
    for month_name, month_num in month_map.items():
        if month_name in col_str:
            month = month_num
            break
    
    # If we have both, create a date
    if year and month:
        try:
            return datetime(year, month, 1)
        except:
            return None
    
    return None

# Create time series dataset
time_series_df = create_timeseries_dataset(df, fund_meta_df, time_period_cols, data_start_idx)
````

````python
# ENHANCED FINANCIAL METRICS CALCULATION
print("\n4. CALCULATING ENHANCED FINANCIAL METRICS")
print("=" * 50)

def calculate_enhanced_metrics(time_series_df):
    """Calculate comprehensive financial metrics with proper time series"""
    
    if len(time_series_df) == 0:
        print("No time series data available")
        return pd.DataFrame()
    
    # Sort by date for proper time series analysis
    time_series_df = time_series_df.sort_values(['fund_name', 'date'])
    
    results = []
    
    for fund_name in time_series_df['fund_name'].unique():
        fund_data = time_series_df[time_series_df['fund_name'] == fund_name]
        
        if len(fund_data) < 2:
            continue
            
        # Calculate monthly returns
        fund_data = fund_data.copy()
        fund_data['monthly_return'] = fund_data['nav_value'].pct_change()
        
        returns = fund_data['monthly_return'].dropna()
        
        if len(returns) < 2:
            continue
            
        try:
            # Basic return metrics
            total_months = len(returns)
            total_return = (returns + 1).prod() - 1
            cagr = (1 + total_return) ** (12 / total_months) - 1
            avg_monthly_return = returns.mean()
            
            # Risk metrics
            monthly_volatility = returns.std()
            annualized_volatility = monthly_volatility * np.sqrt(12)
            
            # Maximum Drawdown
            cumulative_returns = (1 + returns).cumprod()
            running_max = cumulative_returns.expanding().max()
            drawdown = (cumulative_returns - running_max) / running_max
            max_drawdown = drawdown.min()
            
            # VaR (95% confidence)
            var_95 = np.percentile(returns, 5)
            
            # Downside deviation
            downside_returns = returns[returns < 0]
            downside_deviation = downside_returns.std() if len(downside_returns) > 0 else 0
            
            # Risk-adjusted returns
            risk_free_rate = 0.10  # 10% for Nigeria
            sharpe_ratio = (cagr - risk_free_rate) / annualized_volatility if annualized_volatility != 0 else 0
            sortino_ratio = (cagr - risk_free_rate) / (downside_deviation * np.sqrt(12)) if downside_deviation != 0 else 0
            
            # Additional metrics
            positive_months = (returns > 0).sum()
            negative_months = (returns < 0).sum()
            win_rate = positive_months / len(returns)
            
            fund_info = fund_data.iloc[0]
            results.append({
                'fund_manager': fund_info['fund_manager'],
                'fund_name': fund_name,
                'observation_months': total_months,
                'total_return': total_return,
                'cagr': cagr,
                'avg_monthly_return': avg_monthly_return,
                'monthly_volatility': monthly_volatility,
                'annualized_volatility': annualized_volatility,
                'max_drawdown': max_drawdown,
                'var_95': var_95,
                'sharpe_ratio': sharpe_ratio,
                'sortino_ratio': sortino_ratio,
                'win_rate': win_rate,
                'positive_months': positive_months,
                'negative_months': negative_months,
                'data_points': len(fund_data)
            })
            
        except Exception as e:
            print(f"Error processing {fund_name}: {e}")
            continue
    
    metrics_df = pd.DataFrame(results)
    print(f"Calculated metrics for {len(metrics_df)} funds")
    
    if len(metrics_df) > 0:
        print("\nPerformance Summary:")
        summary = metrics_df[['cagr', 'annualized_volatility', 'sharpe_ratio', 'max_drawdown']].describe()
        print(summary.round(4))
    
    return metrics_df

metrics_df = calculate_enhanced_metrics(time_series_df)
````

````python
# CORRECTED APPROACH FOR PORTFOLIO COMPOSITION DATA
print("=== PROCESSING PORTFOLIO COMPOSITION DATA ===")
print("=" * 50)

def extract_portfolio_data_corrected(df):
    """Extract fund portfolio composition data correctly"""
    
    print("1. IDENTIFYING FUND DATA PATTERNS")
    print("-" * 40)
    
    # We can see the pattern: 
    # - Row 1: Header
    # - Row 2: "EXCHANGE TRADED FUNDS" 
    # - Row 3+: Actual fund data with manager, name, and portfolio values
    
    fund_data_rows = []
    
    for idx in range(len(df)):
        row = df.iloc[idx]
        
        # Look for rows with fund manager names (like "Lotus Capital Limited")
        if (pd.notna(row.iloc[1]) and 
            isinstance(row.iloc[1], str) and
            'limited' in row.iloc[1].lower() and  # Fund managers typically have "Limited"
            'fund' not in row.iloc[1].lower() and  # Exclude header rows
            'manager' not in row.iloc[1].lower()):
            
            fund_info = {
                'row_index': idx,
                'fund_manager': row.iloc[1],
                'fund_name': row.iloc[2] if len(row) > 2 and pd.notna(row.iloc[2]) else "Unknown",
                'equities': pd.to_numeric(row.iloc[3], errors='coerce') if len(row) > 3 else None,
                'money_market': pd.to_numeric(row.iloc[4], errors='coerce') if len(row) > 4 else None,
                'bonds': pd.to_numeric(row.iloc[5], errors='coerce') if len(row) > 5 else None,
                'total_investment': pd.to_numeric(row.iloc[6], errors='coerce') if len(row) > 6 else None,
                'cash_balance': pd.to_numeric(row.iloc[7], errors='coerce') if len(row) > 7 else None,
                'total_assets': pd.to_numeric(row.iloc[8], errors='coerce') if len(row) > 8 else None,
                'total_liabilities': pd.to_numeric(row.iloc[9], errors='coerce') if len(row) > 9 else None
            }
            fund_data_rows.append(fund_info)
    
    print(f"Found {len(fund_data_rows)} fund data rows")
    
    # Create portfolio dataframe
    portfolio_df = pd.DataFrame(fund_data_rows)
    
    if len(portfolio_df) > 0:
        print("\nSample portfolio data:")
        print(portfolio_df[['fund_manager', 'fund_name', 'total_assets']].head())
    
    return portfolio_df

# Extract portfolio data
portfolio_df = extract_portfolio_data_corrected(df)
````

````python
print("=== COMPREHENSIVE FUND ANALYSIS ===")

def analyze_fund_universe(portfolio_df):
    """Comprehensive analysis of the fund universe"""
    
    print("1. FUND UNIVERSE OVERVIEW")
    print("=" * 50)
    
    print(f"Total fund observations: {len(portfolio_df):,}")
    print(f"Unique funds: {portfolio_df['fund_name'].nunique()}")
    print(f"Unique fund managers: {portfolio_df['fund_manager'].nunique()}")
    
    # Top fund managers by number of funds
    print(f"\n2. FUND MANAGER ANALYSIS")
    manager_stats = portfolio_df.groupby('fund_manager').agg({
        'fund_name': 'nunique',
        'total_assets': ['sum', 'mean', 'max']
    }).round(2)
    
    manager_stats.columns = ['num_funds', 'total_aum', 'avg_fund_size', 'largest_fund']
    manager_stats = manager_stats.sort_values('total_aum', ascending=False)
    
    print("Top 10 Fund Managers by AUM:")
    print(manager_stats.head(10))
    
    # Largest funds
    print(f"\n3. LARGEST FUNDS")
    largest_funds = portfolio_df.groupby('fund_name').agg({
        'fund_manager': 'first',
        'total_assets': 'max'
    }).nlargest(10, 'total_assets')
    
    for idx, (fund_name, row) in enumerate(largest_funds.iterrows(), 1):
        print(f"{idx:2d}. {fund_name:30} {row['fund_manager'][:20]:20} ₦{row['total_assets']:,.0f}")
    
    return manager_stats, largest_funds

# Run comprehensive analysis
manager_stats, largest_funds = analyze_fund_universe(portfolio_df)
````

````python
# ASSET DISTRIBUTION ANALYSIS
print("\n4. ASSET DISTRIBUTION ANALYSIS")
print("=" * 50)

def analyze_asset_distribution(portfolio_df):
    """Analyze how assets are distributed across funds"""
    
    import matplotlib.pyplot as plt
    import numpy as np
    
    # Asset size categories
    asset_ranges = [0, 1e6, 1e7, 1e8, 1e9, 1e10]  # 0, 1M, 10M, 100M, 1B, 10B
    asset_labels = ['<₦1M', '₦1-10M', '₦10-100M', '₦100M-1B', '>₦1B']
    
    portfolio_df['asset_category'] = pd.cut(portfolio_df['total_assets'], 
                                          bins=asset_ranges, 
                                          labels=asset_labels)
    
    category_counts = portfolio_df['asset_category'].value_counts()
    
    print("Fund Distribution by Asset Size:")
    for category, count in category_counts.items():
        percentage = (count / len(portfolio_df)) * 100
        print(f"  {category}: {count:3d} funds ({percentage:.1f}%)")
    
    # Visualization
    plt.figure(figsize=(12, 8))
    
    # Pie chart of asset distribution
    plt.subplot(2, 2, 1)
    plt.pie(category_counts.values, labels=category_counts.index, autopct='%1.1f%%')
    plt.title('Fund Distribution by Asset Size')
    
    # Top 20 funds by size
    plt.subplot(2, 2, 2)
    top_20 = portfolio_df.nlargest(20, 'total_assets')
    plt.barh(range(len(top_20)), top_20['total_assets'] / 1e6)  # Convert to millions
    plt.yticks(range(len(top_20)), [name[:15] + '...' for name in top_20['fund_name']], fontsize=8)
    plt.xlabel('Total Assets (₦ Millions)')
    plt.title('Top 20 Largest Funds')
    
    # Fund manager concentration
    plt.subplot(2, 2, 3)
    top_managers = portfolio_df.groupby('fund_manager')['total_assets'].sum().nlargest(10)
    plt.bar(range(len(top_managers)), top_managers.values / 1e6)  # Convert to millions
    plt.xticks(range(len(top_managers)), [name[:10] + '...' for name in top_managers.index], 
               rotation=45, fontsize=8)
    plt.ylabel('Total AUM (₦ Millions)')
    plt.title('Top 10 Fund Managers by AUM')
    
    # Asset size histogram
    plt.subplot(2, 2, 4)
    plt.hist(portfolio_df['total_assets'] / 1e6, bins=50, alpha=0.7, edgecolor='black')
    plt.xlabel('Total Assets (₦ Millions)')
    plt.ylabel('Number of Funds')
    plt.title('Distribution of Fund Sizes')
    plt.yscale('log')  # Log scale for better visualization
    
    plt.tight_layout()
    plt.show()
    
    return category_counts

# Run distribution analysis
category_counts = analyze_asset_distribution(portfolio_df)
````

````python
print("=== COMPREHENSIVE ETF MARKET STRUCTURE ANALYSIS ===")
print("=" * 60)

def comprehensive_market_analysis(portfolio_df):
    """Deep analysis of Nigerian ETF market structure"""
    
    print("1. MARKET CONCENTRATION ANALYSIS")
    print("=" * 50)
    
    # Calculate Herfindahl-Hirschman Index (HHI) for market concentration
    total_market_aum = portfolio_df['total_assets'].sum()
    print(f"Total ETF Market AUM: ₦{total_market_aum:,.0f}")
    
    # Market share by manager
    manager_market_share = portfolio_df.groupby('fund_manager')['total_assets'].sum() / total_market_aum * 100
    hhi = (manager_market_share ** 2).sum()
    
    print(f"Market Concentration (HHI): {hhi:.0f}")
    if hhi > 2500:
        concentration_level = "Highly Concentrated"
    elif hhi > 1500:
        concentration_level = "Moderately Concentrated" 
    else:
        concentration_level = "Unconcentrated"
    print(f"Market Structure: {concentration_level}")
    
    print(f"\nTop 5 Managers Market Share:")
    top_5_share = manager_market_share.nlargest(5)
    for manager, share in top_5_share.items():
        print(f"  {manager[:25]:25} {share:5.1f}%")
    
    # CR3 and CR5 concentration ratios
    cr3 = top_5_share.head(3).sum()
    cr5 = top_5_share.head(5).sum()
    print(f"CR3 Concentration: {cr3:.1f}%")
    print(f"CR5 Concentration: {cr5:.1f}%")
    
    return manager_market_share, hhi

manager_market_share, hhi = comprehensive_market_analysis(portfolio_df)
````

````python
# ASSET ALLOCATION ANALYSIS
print("\n2. ASSET ALLOCATION ANALYSIS")
print("=" * 50)

def analyze_asset_allocation(portfolio_df):
    """Analyze how ETFs allocate assets across classes"""
    
    # Calculate portfolio composition percentages
    portfolio_df['equities_pct'] = (portfolio_df['equities'] / portfolio_df['total_assets'] * 100).fillna(0)
    portfolio_df['money_market_pct'] = (portfolio_df['money_market'] / portfolio_df['total_assets'] * 100).fillna(0)
    portfolio_df['bonds_pct'] = (portfolio_df['bonds'] / portfolio_df['total_assets'] * 100).fillna(0)
    portfolio_df['cash_pct'] = (portfolio_df['cash_balance'] / portfolio_df['total_assets'] * 100).fillna(0)
    
    print("Overall Asset Allocation Averages:")
    print(f"  Equities:    {portfolio_df['equities_pct'].mean():5.1f}%")
    print(f"  Money Market: {portfolio_df['money_market_pct'].mean():5.1f}%") 
    print(f"  Bonds:        {portfolio_df['bonds_pct'].mean():5.1f}%")
    print(f"  Cash:         {portfolio_df['cash_pct'].mean():5.1f}%")
    
    # Categorize funds by primary asset class
    def categorize_fund(row):
        allocations = {
            'equities': row['equities_pct'],
            'money_market': row['money_market_pct'], 
            'bonds': row['bonds_pct'],
            'cash': row['cash_pct']
        }
        primary_class = max(allocations, key=allocations.get)
        return primary_class if allocations[primary_class] > 50 else 'mixed'
    
    portfolio_df['primary_asset_class'] = portfolio_df.apply(categorize_fund, axis=1)
    
    class_distribution = portfolio_df['primary_asset_class'].value_counts()
    print(f"\nFund Distribution by Primary Asset Class:")
    for asset_class, count in class_distribution.items():
        pct = (count / len(portfolio_df)) * 100
        print(f"  {asset_class:12}: {count:3d} funds ({pct:.1f}%)")
    
    return portfolio_df

portfolio_df = analyze_asset_allocation(portfolio_df)
````

````python
# FUND EFFICIENCY AND LEVERAGE ANALYSIS
print("\n3. FUND EFFICIENCY ANALYSIS")
print("=" * 50)

def analyze_fund_efficiency(portfolio_df):
    """Analyze fund efficiency metrics"""
    
    # Calculate expense ratio proxy (liabilities/assets)
    portfolio_df['expense_ratio_proxy'] = (portfolio_df['total_liabilities'] / portfolio_df['total_assets'] * 100).fillna(0)
    
    # Calculate cash drag (high cash as % of assets)
    portfolio_df['cash_drag'] = portfolio_df['cash_pct']
    
    # Investment efficiency (total investment / total assets)
    portfolio_df['investment_efficiency'] = (portfolio_df['total_investment'] / portfolio_df['total_assets'] * 100).fillna(0)
    
    print("Fund Efficiency Metrics:")
    print(f"  Average Expense Ratio Proxy: {portfolio_df['expense_ratio_proxy'].mean():.2f}%")
    print(f"  Average Cash Drag: {portfolio_df['cash_drag'].mean():.1f}%")
    print(f"  Average Investment Efficiency: {portfolio_df['investment_efficiency'].mean():.1f}%")
    
    # Identify most efficient funds
    efficient_funds = portfolio_df[
        (portfolio_df['expense_ratio_proxy'] < portfolio_df['expense_ratio_proxy'].median()) &
        (portfolio_df['investment_efficiency'] > portfolio_df['investment_efficiency'].median())
    ].nlargest(10, 'total_assets')
    
    print(f"\nTop 10 Most Efficient Large Funds:")
    for idx, row in efficient_funds.iterrows():
        print(f"  {row['fund_name'][:30]:30} AUM: ₦{row['total_assets']:,.0f}")
    
    return portfolio_df

portfolio_df = analyze_fund_efficiency(portfolio_df)
````

````python
# PEER GROUP ANALYSIS AND BENCHMARKING
print("\n4. PEER GROUP ANALYSIS")
print("=" * 50)

def peer_group_analysis(portfolio_df):
    """Compare funds within peer groups"""
    
    # Create peer groups by asset class and size
    size_breaks = portfolio_df['total_assets'].quantile([0.33, 0.67])
    small_cap = size_breaks.iloc[0]
    large_cap = size_breaks.iloc[1]
    
    def size_category(assets):
        if assets <= small_cap:
            return 'Small'
        elif assets <= large_cap:
            return 'Medium' 
        else:
            return 'Large'
    
    portfolio_df['size_category'] = portfolio_df['total_assets'].apply(size_category)
    
    print("Peer Group Analysis:")
    peer_groups = portfolio_df.groupby(['primary_asset_class', 'size_category'])
    
    for (asset_class, size), group in peer_groups:
        if len(group) > 0:
            avg_assets = group['total_assets'].mean()
            avg_expense = group['expense_ratio_proxy'].mean()
            avg_efficiency = group['investment_efficiency'].mean()
            
            print(f"\n  {asset_class} - {size} Funds ({len(group)} funds):")
            print(f"    Avg AUM: ₦{avg_assets:,.0f}")
            print(f"    Avg Expense: {avg_expense:.2f}%")
            print(f"    Avg Efficiency: {avg_efficiency:.1f}%")
    
    # Identify outliers (potential opportunities or risks)
    high_expense_funds = portfolio_df[portfolio_df['expense_ratio_proxy'] > portfolio_df['expense_ratio_proxy'].quantile(0.9)]
    low_efficiency_funds = portfolio_df[portfolio_df['investment_efficiency'] < portfolio_df['investment_efficiency'].quantile(0.1)]
    
    print(f"\nOutlier Analysis:")
    print(f"  High Expense Funds (>90th percentile): {len(high_expense_funds)}")
    print(f"  Low Efficiency Funds (<10th percentile): {len(low_efficiency_funds)}")
    
    return portfolio_df

portfolio_df = peer_group_analysis(portfolio_df)
````

````python
print("=== DATA QUALITY CLEANING ===")
print("=" * 50)

def clean_data_quality_issues(portfolio_df):
    """Fix the serious data quality issues"""
    
    print("BEFORE CLEANING - Data Issues:")
    print(f"Total records: {len(portfolio_df)}")
    
    # Check for zero or negative assets (causes infinite ratios)
    zero_assets = (portfolio_df['total_assets'] <= 0).sum()
    print(f"Records with zero/negative assets: {zero_assets}")
    
    # Check for missing values
    missing_assets = portfolio_df['total_assets'].isna().sum()
    print(f"Records with missing assets: {missing_assets}")
    
    # Remove problematic records
    clean_df = portfolio_df.copy()
    
    # 1. Remove records with invalid asset values
    clean_df = clean_df[clean_df['total_assets'] > 1000]  # Reasonable minimum for ETFs
    
    # 2. Remove records where liabilities > assets (impossible)
    clean_df = clean_df[clean_df['total_liabilities'] <= clean_df['total_assets']]
    
    # 3. Remove records with missing critical data
    clean_df = clean_df.dropna(subset=['fund_manager', 'fund_name', 'total_assets'])
    
    print(f"\nAFTER CLEANING:")
    print(f"Remaining records: {len(clean_df)}")
    print(f"Records removed: {len(portfolio_df) - len(clean_df)}")
    
    return clean_df

# Clean the data
clean_portfolio = clean_data_quality_issues(portfolio_df)
````

````python
print("=== ANALYZING CLEAN ETF DATASET ===")
print("=" * 50)

def analyze_clean_dataset(clean_df):
    """Comprehensive analysis of the cleaned ETF data"""
    
    print(f"WORKING WITH {len(clean_df)} CLEAN RECORDS")
    print("=" * 40)
    
    # Basic overview
    total_aum = clean_df['total_assets'].sum()
    avg_fund_size = clean_df['total_assets'].mean()
    
    print(f"Total ETF Market AUM: ₦{total_aum:,.0f}")
    print(f"Average Fund Size: ₦{avg_fund_size:,.0f}")
    print(f"Number of Fund Managers: {clean_df['fund_manager'].nunique()}")
    print(f"Number of Unique Funds: {clean_df['fund_name'].nunique()}")
    
    # Market concentration analysis
    manager_aum = clean_df.groupby('fund_manager')['total_assets'].sum().sort_values(ascending=False)
    
    print(f"\nTOP 10 FUND MANAGERS BY AUM:")
    print("-" * 50)
    for i, (manager, aum) in enumerate(manager_aum.head(10).items(), 1):
        market_share = (aum / total_aum) * 100
        print(f"{i:2d}. {manager[:35]:35} ₦{aum:>12,.0f} ({market_share:.1f}%)")
    
    # Calculate HHI for market concentration
    market_shares = (manager_aum / total_aum * 100)
    hhi = (market_shares ** 2).sum()
    
    print(f"\nMARKET CONCENTRATION:")
    print(f"Herfindahl Index (HHI): {hhi:,.0f}")
    if hhi > 2500:
        print("Market: HIGHLY CONCENTRATED")
    elif hhi > 1500:
        print("Market: MODERATELY CONCENTRATED")
    else:
        print("Market: COMPETITIVE")
    
    # CR3 and CR5 ratios
    cr3 = market_shares.head(3).sum()
    cr5 = market_shares.head(5).sum()
    print(f"CR3 Concentration: {cr3:.1f}%")
    print(f"CR5 Concentration: {cr5:.1f}%")
    
    return manager_aum, hhi

manager_aum, hhi = analyze_clean_dataset(clean_portfolio)
````

````python
# FUND SIZE DISTRIBUTION ANALYSIS
print("\n=== FUND SIZE DISTRIBUTION ===")
print("=" * 50)

def analyze_size_distribution(clean_df):
    """Analyze the distribution of fund sizes"""
    
    import matplotlib.pyplot as plt
    import numpy as np
    
    # Create realistic size categories
    size_bins = [0, 100e6, 500e6, 1e9, 5e9, np.inf]  # 0-100M, 100-500M, 500M-1B, 1-5B, 5B+
    size_labels = ['<₦100M', '₦100-500M', '₦500M-1B', '₦1-5B', '>₦5B']
    
    clean_df['size_bracket'] = pd.cut(clean_df['total_assets'], bins=size_bins, labels=size_labels)
    
    size_distribution = clean_df['size_bracket'].value_counts().sort_index()
    
    print("FUND SIZE DISTRIBUTION:")
    for bracket, count in size_distribution.items():
        percentage = (count / len(clean_df)) * 100
        avg_size = clean_df[clean_df['size_bracket'] == bracket]['total_assets'].mean()
        print(f"  {bracket:12}: {count:2d} funds ({percentage:4.1f}%) - Avg: ₦{avg_size:,.0f}")
    
    # Largest funds
    print(f"\nLARGEST 10 FUNDS:")
    largest_funds = clean_df.nlargest(10, 'total_assets')[['fund_name', 'fund_manager', 'total_assets']]
    for idx, row in largest_funds.iterrows():
        print(f"  {row['fund_name'][:30]:30} {row['fund_manager'][:20]:20} ₦{row['total_assets']:>12,.0f}")
    
    # Visualization
    plt.figure(figsize=(12, 8))
    
    # Fund size distribution
    plt.subplot(2, 2, 1)
    plt.hist(clean_df['total_assets'] / 1e6, bins=20, alpha=0.7, edgecolor='black')
    plt.xlabel('Total Assets (₦ Millions)')
    plt.ylabel('Number of Funds')
    plt.title('Distribution of Fund Sizes')
    
    # Size bracket pie chart
    plt.subplot(2, 2, 2)
    plt.pie(size_distribution.values, labels=size_distribution.index, autopct='%1.1f%%')
    plt.title('Fund Size Distribution')
    
    # Top managers by AUM
    plt.subplot(2, 2, 3)
    top_managers = manager_aum.head(10)
    plt.barh(range(len(top_managers)), top_managers.values / 1e6)
    plt.yticks(range(len(top_managers)), [name[:15] + '...' for name in top_managers.index])
    plt.xlabel('Total AUM (₦ Millions)')
    plt.title('Top 10 Fund Managers by AUM')
    
    # Box plot of fund sizes by manager (top 8 only)
    plt.subplot(2, 2, 4)
    top_8_managers = manager_aum.head(8).index
    size_data = []
    labels = []
    for manager in top_8_managers:
        manager_funds = clean_df[clean_df['fund_manager'] == manager]['total_assets'] / 1e6
        if len(manager_funds) > 0:
            size_data.append(manager_funds)
            labels.append(manager[:10] + '...')
    
    if size_data:
        plt.boxplot(size_data, labels=labels)
        plt.xticks(rotation=45)
        plt.ylabel('Fund Size (₦ Millions)')
        plt.title('Fund Size Distribution by Manager')
    
    plt.tight_layout()
    plt.show()
    
    return size_distribution

size_distribution = analyze_size_distribution(clean_portfolio)
````

````python
# ASSET ALLOCATION ANALYSIS
print("\n=== ASSET ALLOCATION ANALYSIS ===")
print("=" * 50)

def analyze_asset_allocation(clean_df):
    """Analyze how ETFs allocate their assets"""
    
    print("OVERALL ASSET ALLOCATION:")
    avg_equities = clean_df['equities_pct'].mean()
    avg_bonds = clean_df['bonds_pct'].mean()
    avg_money_market = clean_df['money_market_pct'].mean()
    avg_cash = clean_df['cash_pct'].mean()
    
    print(f"  Equities:    {avg_equities:5.1f}%")
    print(f"  Bonds:        {avg_bonds:5.1f}%")
    print(f"  Money Market: {avg_money_market:5.1f}%")
    print(f"  Cash:         {avg_cash:5.1f}%")
    
    # Categorize funds by primary focus
    def get_primary_asset(row):
        allocations = {
            'Equities': row['equities_pct'],
            'Bonds': row['bonds_pct'],
            'Money Market': row['money_market_pct'],
            'Cash': row['cash_pct']
        }
        return max(allocations, key=allocations.get)
    
    clean_df['primary_asset'] = clean_df.apply(get_primary_asset, axis=1)
    
    primary_asset_dist = clean_df['primary_asset'].value_counts()
    print(f"\nFUNDS BY PRIMARY ASSET CLASS:")
    for asset, count in primary_asset_dist.items():
        percentage = (count / len(clean_df)) * 100
        avg_size = clean_df[clean_df['primary_asset'] == asset]['total_assets'].mean()
        print(f"  {asset:12}: {count:2d} funds ({percentage:4.1f}%) - Avg AUM: ₦{avg_size:,.0f}")
    
    # Expense and efficiency analysis
    print(f"\nCOST AND EFFICIENCY ANALYSIS:")
    print(f"  Average Expense Ratio: {clean_df['expense_ratio_proxy'].mean():.2f}%")
    print(f"  Average Investment Efficiency: {clean_df['investment_efficiency'].mean():.1f}%")
    
    # Most efficient large funds
    large_funds = clean_df[clean_df['total_assets'] > clean_df['total_assets'].median()]
    efficient_large = large_funds.nlargest(5, 'investment_efficiency')
    
    print(f"\nMOST EFFICIENT LARGE FUNDS:")
    for idx, row in efficient_large.iterrows():
        print(f"  {row['fund_name'][:25]:25} Eff: {row['investment_efficiency']:5.1f}% Cost: {row['expense_ratio_proxy']:4.1f}%")
    
    return clean_df

clean_portfolio = analyze_asset_allocation(clean_portfolio)
````

````python
# 3. RESEARCH VISUALIZATION PACKAGE
print("\n3. CREATING RESEARCH VISUALIZATIONS")
print("-" * 40)

def create_research_visualizations(clean_portfolio, market_share):
    """Create comprehensive visualizations for research"""
    
    import matplotlib.pyplot as plt
    import seaborn as sns
    
    # Set professional style
    plt.style.use('seaborn-v0_8')
    sns.set_palette("husl")
    
    # Create a complete visualization dashboard
    fig = plt.figure(figsize=(20, 16))
    
    # 1. Market Share Treemap (simplified as bar chart)
    ax1 = plt.subplot(3, 3, 1)
    top_10_share = market_share.head(10)
    ax1.barh(range(len(top_10_share)), top_10_share.values)
    ax1.set_yticks(range(len(top_10_share)))
    ax1.set_yticklabels([name[:15] + '...' for name in top_10_share.index])
    ax1.set_xlabel('Market Share (%)')
    ax1.set_title('Top 10 Managers by Market Share')
    
    # 2. Fund Size Distribution
    ax2 = plt.subplot(3, 3, 2)
    clean_portfolio['log_assets'] = np.log10(clean_portfolio['total_assets'])
    ax2.hist(clean_portfolio['log_assets'], bins=20, alpha=0.7, edgecolor='black')
    ax2.set_xlabel('Log Total Assets')
    ax2.set_ylabel('Number of Funds')
    ax2.set_title('Fund Size Distribution (Log Scale)')
    
    # 3. Expense vs Efficiency
    ax3 = plt.subplot(3, 3, 3)
    scatter = ax3.scatter(clean_portfolio['expense_ratio_proxy'], 
                         clean_portfolio['investment_efficiency'],
                         c=clean_portfolio['total_assets'], 
                         cmap='viridis', alpha=0.6, s=50)
    plt.colorbar(scatter, ax=ax3, label='Total Assets')
    ax3.set_xlabel('Expense Ratio (%)')
    ax3.set_ylabel('Investment Efficiency (%)')
    ax3.set_title('Cost vs Efficiency Analysis')
    
    # 4. Asset Allocation by Size
    ax4 = plt.subplot(3, 3, 4)
    allocation_by_size = clean_portfolio.groupby('size_bracket').agg({
        'equities_pct': 'mean',
        'bonds_pct': 'mean',
        'money_market_pct': 'mean'
    })
    allocation_by_size.plot(kind='bar', ax=ax4, stacked=True)
    ax4.set_ylabel('Allocation Percentage (%)')
    ax4.set_title('Asset Allocation by Fund Size')
    ax4.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    
    # 5. Manager Concentration
    ax5 = plt.subplot(3, 3, 5)
    cumulative_share = market_share.sort_values(ascending=False).cumsum()
    ax5.plot(range(1, len(cumulative_share) + 1), cumulative_share.values, marker='o')
    ax5.axhline(y=50, color='r', linestyle='--', alpha=0.5, label='50% Market Share')
    ax5.axvline(x=market_share[market_share.cumsum() <= 50].count(), 
                color='r', linestyle='--', alpha=0.5)
    ax5.set_xlabel('Number of Managers')
    ax5.set_ylabel('Cumulative Market Share (%)')
    ax5.set_title('Market Concentration Curve')
    ax5.legend()
    ax5.grid(True, alpha=0.3)
    
    # 6. Size Category Distribution
    ax6 = plt.subplot(3, 3, 6)
    size_counts = clean_portfolio['size_bracket'].value_counts().sort_index()
    ax6.pie(size_counts.values, labels=size_counts.index, autopct='%1.1f%%')
    ax6.set_title('Fund Distribution by Size Category')
    
    # 7. Efficiency by Asset Class
    ax7 = plt.subplot(3, 3, 7)
    efficiency_by_asset = clean_portfolio.groupby('primary_asset')['investment_efficiency'].mean()
    ax7.bar(efficiency_by_asset.index, efficiency_by_asset.values)
    ax7.set_ylabel('Average Investment Efficiency (%)')
    ax7.set_title('Efficiency by Primary Asset Class')
    ax7.tick_params(axis='x', rotation=45)
    
    # 8. Expense Ratio Distribution
    ax8 = plt.subplot(3, 3, 8)
    ax8.hist(clean_portfolio['expense_ratio_proxy'], bins=20, alpha=0.7, edgecolor='black')
    ax8.set_xlabel('Expense Ratio (%)')
    ax8.set_ylabel('Number of Funds')
    ax8.set_title('Expense Ratio Distribution')
    
    # 9. AUM vs Efficiency by Manager (Top 8)
    ax9 = plt.subplot(3, 3, 9)
    top_managers = market_share.head(8).index
    for manager in top_managers:
        manager_data = clean_portfolio[clean_portfolio['fund_manager'] == manager]
        if len(manager_data) > 0:
            ax9.scatter(manager_data['total_assets'].mean() / 1e6,
                       manager_data['investment_efficiency'].mean(),
                       s=manager_data['total_assets'].sum() / 1e6 / 10,  # Size by total AUM
                       label=manager[:12] + '...', alpha=0.7)
    
    ax9.set_xlabel('Average Fund Size (₦ Millions)')
    ax9.set_ylabel('Average Efficiency (%)')
    ax9.set_title('Manager Performance: Size vs Efficiency')
    ax9.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=8)
    
    plt.tight_layout()
    plt.savefig('Comprehensive_ETF_Research_Dashboard.png', dpi=300, bbox_inches='tight')
    plt.show()
    
    print("Comprehensive research dashboard created")

create_research_visualizations(clean_portfolio, market_share)
````
